```python
#Task 1: Pandas Series Basics
#1. Import pandas as pd.
#2. Create a Pandas Series from the list:
#3. marks = [78, 85, 90, 66, 72]
#4. Print:
#• Series values
#• Index
#• Data type
#5. Access:
#• First element
#• Last two elements
import pandas as pd
# Step 1: Import pandas as pd
# Step 2: Create a Pandas Series from the list
marks = [78, 85, 90, 66, 72]
marks_series = pd.Series(marks)
# Step 4: Print Series values, Index, and Data type
print("Series values:")
print(marks_series.values)
print("\nIndex:")
print(marks_series.index)
print("\nData type:")
print(marks_series.dtype)
# Step 5: Access first element and last two elements
print("\nFirst element:")
print(marks_series[0])
print("\nLast two elements:")
print(marks_series[-2:])
```

```python
#Task 2: Mathematical Operations on Series
#Using the marks Series:
#1. Add 5 grace marks to all students.
#2. Subtract 2 marks from all values.
#3. Multiply all marks by 1.05.
#4. Divide all marks by 2.
#Print results for each operation.
# Step 1: Add 5 grace marks to all students
marks_with_grace = marks_series + 5
print("Marks after adding 5 grace marks:")
print(marks_with_grace)
# Step 2: Subtract 2 marks from all values
marks_subtracted = marks_series - 2
print("\nMarks after subtracting 2 marks:")
print(marks_subtracted)
# Step 3: Multiply all marks by 1.05
marks_multiplied = marks_series * 1.05
print("\nMarks after multiplying by 1.05:")
print(marks_multiplied)
# Step 4: Divide all marks by 2
marks_divided = marks_series / 2
print("\nMarks after dividing by 2:")
print(marks_divided)
```

```python
#Task 3: Python Functionalities on Series
#Using the same Series:
#1. Find:
#• Maximum marks
#• Minimum marks
#o Sum of marks
#• Mean marks
#2. Apply a lambda function to check whether each student has passed (>= 70).
#3. Count how many students passed.
# Step 1: Find maximum, minimum, sum, and mean marks
max_marks = marks_series.max()
min_marks = marks_series.min()
sum_marks = marks_series.sum()
mean_marks = marks_series.mean()
print("Maximum marks:", max_marks)
print("Minimum marks:", min_marks)
print("Sum of marks:", sum_marks)
print("Mean marks:", mean_marks)
# Step 2: Apply a lambda function to check whether each student has passed (>= 70)
passed = marks_series.apply(lambda x: x >= 70)
print("\nPassed status for each student:")
print(passed)
# Step 3: Count how many students passed
num_passed = passed.sum()
print("\nNumber of students passed:", num_passed)
```

```python
#Task 4: Create a DataFrame
#Create a DataFrame students:
#students ={
#'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
#'Marks': [78, 85, 90, 66, 72],
#'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
#}
#1. Convert it into a DataFrame.
#2. Print first 3 rows.
#3. Print last 2 rows.
#4. Print DataFrame shape and column names.
# Step 1: Create a DataFrame from the dictionary
students = {
    'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
    'Marks': [78, 85, 90, 66, 72],
    'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
}
students_df = pd.DataFrame(students)
# Step 2: Print first 3 rows
print("First 3 rows of the DataFrame:")
print(students_df.head(3))
# Step 3: Print last 2 rows
print("\nLast 2 rows of the DataFrame:")
print(students_df.tail(2))
# Step 4: Print DataFrame shape and column names
print("\nDataFrame shape:", students_df.shape)
print("Column names:", students_df.columns)
```

```python
#Task 5: Important DataFrame Functions
#Using the students DataFrame:
#1. Use and print output of:
#o .info()
#o .describe()
#o .head()
#o .tail)
#2. Sort students by Marks in descending order.
#3. Reset index after sorting.
# Step 1: Use and print output of .info(), .describe(), .head(), and .tail()
print("DataFrame info:")
print(students_df.info())
print("\nDataFrame description:")
print(students_df.describe())
print("\nFirst 5 rows of the DataFrame:")
print(students_df.head())
print("\nLast 5 rows of the DataFrame:")
print(students_df.tail())
# Step 2: Sort students by Marks in descending order
sorted_students = students_df.sort_values(by='Marks', ascending=False)
print("\nStudents sorted by Marks in descending order:")
print(sorted_students)
# Step 3: Reset index after sorting
sorted_students_reset = sorted_students.reset_index(drop=True)
print("\nSorted students with reset index:")
print(sorted_students_reset)
```

```python
#Task 6: Filtering & Conditional Selection
#Perform the following filters:
#1. Students who scored more than 75 marks.
#2. Students belonging to subject Math.
#3. Students who scored more than average marks.
#4. Students who failed (marks < 70).
# Step 1: Filter students who scored more than 75 marks
students_above_75 = students_df[students_df['Marks'] > 75]
print("Students who scored more than 75 marks:")
print(students_above_75)
# Step 2: Filter students belonging to subject Math
students_math = students_df[students_df['Subject'] == 'Math']
print("\nStudents belonging to subject Math:")
print(students_math)
# Step 3: Filter students who scored more than average marks
average_marks = students_df['Marks'].mean()
students_above_average = students_df[students_df['Marks'] > average_marks]
print("\nStudents who scored more than average marks:")
print(students_above_average)
# Step 4: Filter students who failed (marks < 70)
students_failed = students_df[students_df['Marks'] < 70]
print("\nStudents who failed (marks < 70):")
print(students_failed)  
```

```python
#Task 7: Grouping & Basic Analysis
#1. Find average marks per subject using groupby().
#2. Count number of students per subject.
#3. Find maximum marks per subject.
# Step 1: Find average marks per subject using groupby()
average_marks_per_subject = students_df.groupby('Subject')['Marks'].mean()
print("Average marks per subject:")
print(average_marks_per_subject)
# Step 2: Count number of students per subject
students_per_subject = students_df.groupby('Subject')['Name'].count()
print("\nNumber of students per subject:")
print(students_per_subject)
# Step 3: Find maximum marks per subject
max_marks_per_subject = students_df.groupby('Subject')['Marks'].max()
print("\nMaximum marks per subject:")
print(max_marks_per_subject)
```

```python
#Task 9: Mini Use Case: Sales Data Analysis
#Create a DataFrame:
#sales = {
#'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri],
#'Revenue': [1200, 1500, 900, 2000, 1800]
#}
#Perform:
#1. Total revenue
#2. Average daily revenue
#3. Day with highest revenue
#4. Days where revenue › average
#5. Plot revenue vs day
# Step 1: Create a DataFrame from the dictionary
sales = {
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Revenue': [1200, 1500, 900, 2000, 1800]
}
sales_df = pd.DataFrame(sales)
# Step 2: Total revenue
total_revenue = sales_df['Revenue'].sum()
print("Total revenue:", total_revenue)
# Step 3: Average daily revenue
average_revenue = sales_df['Revenue'].mean()
print("Average daily revenue:", average_revenue)
# Step 4: Day with highest revenue
day_highest_revenue = sales_df.loc[sales_df['Revenue'].idxmax(), 'Day']
print("Day with highest revenue:", day_highest_revenue)
# Step 5: Days where revenue > average
days_above_average = sales_df[sales_df['Revenue'] > average_revenue]['Day']
print("Days where revenue > average:", days_above_average.tolist())
# Step 6: Plot revenue vs day
sales_df.plot(x='Day', y='Revenue', kind='bar', title='Revenue vs Day')
```
